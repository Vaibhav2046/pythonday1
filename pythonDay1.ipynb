```python
# SEPARATOR


print("vaibhav""pandey") # yarr! there is no space between arguments after running BUT i want space between arguments       /// output: vaibhavpandey

# i got an idea let's see what will happen if i give space between  arguments
print("vaibhav" "pandey") # output is same like- print("vaibhav""pandey") /// output: vaibhavpandey

print("vaibhav","pandey") #  /// output: vaibhav pandey
# DEFAULT sparator is "space"

# got another idea, see BELOW :)
# print("vaibhav"-"pandey")---  showing syntax error :(

# ALL DOUBT CLEAR
print("vaibhav","pandey",sep=",") # output: vaibhav,pandey
print("vaibhav""pandey",sep=",") # we didn't used separator(,) so, no means of assing any value to it /// output: vaibhavpandey

# when a user want to separate two argument he must put "," between arguments and here "," is not a "comma" here "," means SEPARATOR which DEFAULT value is "SPACE"
# now if u want to separate arguments by "," u have to give value to separator LIKE [ sep=","] or [ sep="-"]
# but if u will not separate arguments then there is no mean of giving value to "sep" like in this code [print("vaibhav""pandey",sep=",") # output: vaibhavpandey]

#  NOT USELESS THING :(
print("vaibhav","shaktimaan",sep=" hi hai ") # u can give any value to "sep" /// output: vaibhav hi hai shaktimaan
print("u are not","shaktimaan","bro!",sep="😂") # output: u are not😂shaktimaan😂bro!



```
