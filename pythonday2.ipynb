```python
# finding address of a VARIABLE

a=99
b=a
print(id(a),id(b),a+b,a-b,a*b,a/b,sep=("🤷‍♂️"),end=("🦍"))  # id() used to find address of any variable
print("\n")
a=a+1
print(id(a)) # new object new address
```

```python
""" What is module? module can be any file (in python files end with .py) ,Module contains variables,classes,functions,
etc like- math :a python module contains functions of mathematics operations"""
import math
print(math.sqrt(49))
from math import sqrt          # can improt the functions directly
print(sqrt(49))
import math as shaktimaan       #😂
print(shaktimaan.sqrt(49))
```
